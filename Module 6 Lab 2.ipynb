```python
import sqlite3
import pandas as pd

con = sqlite3.connect(":memory:")
```

```python
con.execute("""
CREATE TABLE customers (
    id INTEGER,
    name TEXT,
    city TEXT,
    balance REAL
)
""")
```

```python
con.executemany("""
INSERT INTO customers (id, name, city, balance)
VALUES (?, ?, ?, ?)
""", [
    (1, 'John Smith', 'Los Angeles', 1250.50),
    (2, 'Maria Garcia', 'San Diego', 2300.75),
    (3, 'David Lee', 'San Francisco', 875.25),
    (4, 'Sarah Johnson', 'Sacramento', 3100.00),
    (5, 'Michael Brown', 'Long Beach', 450.50),
    (6, 'Emily Davis', 'Pasadena', 1825.75),
    (7, 'James Wilson', 'Anaheim', 950.00),
    (8, 'Lisa Martinez', 'Irvine', 2750.25)
])
```

```python
con.commit()
```

```python
df = pd.read_sql_query("SELECT * FROM customers", con)
df
```
